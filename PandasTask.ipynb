## Practice 8: Merging and Joining

**Q13.** Create a DataFrame `df_dept` with columns `Department` and `Manager` mapping 'IT' to 'Alice', 'HR' to 'Bob', and 'Finance' to 'Charlie'. Merge it with `df_emp` (from Practice 5) using a left join.

````python
df_dept = pd.DataFrame({'Department': ['IT', 'HR', 'Finance'], 'Manager': ['Alice', 'Bob', 'Charlie']})
merged=pd.merge(df_emp,df_dept,how="left") 
print(merged)
````

## Practice 9: Pivot Tables

**Q14.** Create a pivot table from the merged DataFrame that shows the total (sum) and average (mean) salary for each manager.

````python
pivot=pd.pivot_table(
    merged,
    values="Salary",#Which its average is to be calculated.
    index="Manager",# The variables which are to be displayed as rows.
    aggfunc=["mean","sum"]#the aggregation function that will evaluate the average and the sum 
    )
print(pivot)
````

## Practice 10: Advanced Data Manipulation

**Q15.** Given a DataFrame with sales data, calculate the rolling 3-day average of 'Revenue'.
```python
sales_data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Product': ['A', 'B', 'A', 'B', 'A', 'A', 'B', 'B', 'A', 'B'],
    'Revenue': [100, 200, 150, 250, 300, 250, 100, 150, 200, 300]
}
df_sales = pd.DataFrame(sales_data)
```

````python
sales_data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Product': ['A', 'B', 'A', 'B', 'A', 'A', 'B', 'B', 'A', 'B'],
    'Revenue': [100, 200, 150, 250, 300, 250, 100, 150, 200, 300]
}
df_sales = pd.DataFrame(sales_data)
df_sales["Rolling 3-day Average of Revenue"]=df_sales["Revenue"].rolling(window=3).mean()
print(df_sales)
````

**Q16.** Find the top 2 highest revenue days for each product from `df_sales`.

````python
top_2_revenues_indicies=df_sales.groupby("Product")["Revenue"].nlargest(2).index.get_level_values(1)#treturns the indicies where the top 2 revenue of each product are placed
Top_two=df_sales.loc[top_2_revenues_indicies]#retuns the rows of the top two for each product
Top_two=Top_two.drop("Rolling 3-day Average of Revenue",axis=1)#removing the rolling
print(Top_two)#prints : date, product, revenu 
````

**Q17.** Extract the year and month from the 'Date' column in `df_sales` and create two new columns 'Year' and 'Month'.

````python
Year=df_sales["Date"].dt.year
Month=df_sales["Date"].dt.month_name()
df_sales["Year"]=Year
df_sales["Month"]=Month
print(df_sales)
````
